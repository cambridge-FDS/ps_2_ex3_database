# Insight ① — How often did the home team win?

**Question.** How did the share of NBA regular-season games won by the designated home team change from 1980–81 through 2022–23? Each eligible game contributes once. We compare final scores, restrict to regular-season IDs, and exclude missing game IDs, missing scores and ties. Start Jupyter from the repository root using the `nba` environment, with the shared database in `data/nba.sqlite`.

```python
from contextlib import closing
from pathlib import Path
from IPython.display import display
from nba.data_loader import connect_database
from nba.insight_1 import home_win_trend
from nba.insight_1_viz import plot_home_win_trend

with closing(connect_database(Path("data/nba.sqlite"))) as connection:
    trend = home_win_trend(connection)
display(trend.head())
display(trend.tail())
print(f"Seasons: {len(trend)}, games: {trend.games.sum():,}, home wins: {trend.home_wins.sum():,}")
missing_years = sorted(set(range(int(trend.season_start_year.min()), int(trend.season_start_year.max()) + 1)) - set(trend.season_start_year))
print("Missing season start years:", missing_years)
```

```python
figure = plot_home_win_trend(trend)
display(figure)
```

## Finding and limitations

In the downloaded database, 28,502 of 47,027 eligible games (60.6%) were won by the designated home team across 42 recorded seasons. The seasonal share was 62.1% in 1980–81 and 58.0% in 2022–23; its highest recorded value was 67.9% in both 1983–84 and 1987–88, and its lowest was 54.4% in 2020–21. This is a descriptive historical pattern, not proof that playing at home caused those wins.

The database contains no regular-season games for 2012–13, so the chart leaves a visible break rather than interpolating. Seasons have different numbers of games, and pandemic-era disruptions or neutral-site games can make the designated home side different from a typical home-court setting. Changes in team strength, scheduling and coverage are not controlled for. The 2022–23 season is the database endpoint, not a claim about the current NBA. The SQL is in `queries/insight_1.sql`; the function rejects duplicate eligible game IDs rather than silently counting them twice.
