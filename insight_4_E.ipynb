```python
import pandas as pd
from pathlib import Path
from nba.data_loader import connect_database
from nba.insight_4_e import get_monthly_game_counts, plot_monthly_game_counts

# Run the notebook with the repository root as its working directory.
DATA_PATH = Path.cwd() / 'data' / 'nba.sqlite'

con = connect_database(DATA_PATH)  # open the existing database read-only
query = 'SELECT * FROM game LIMIT 10'  # write a query that selects the first 10 rows from the game table
top_10_game = pd.read_sql(query, con)  # read the query into a pandas dataframe
top_10_game.head()  # print the dataframe
```

```python
query = """
SELECT
    game_date,
    strftime('%m', game_date) AS month
FROM game
WHERE game_date IS NOT NULL
LIMIT 10
"""

display(pd.read_sql(query, con))
```

```python
query = """
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT game_date) AS distinct_dates,
    COUNT(strftime('%m', game_date)) AS valid_month_rows
FROM game
"""

display(pd.read_sql(query, con))
```

```python
monthly_games = get_monthly_game_counts(con)

display(monthly_games)
```
