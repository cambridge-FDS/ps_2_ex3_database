# Insight ② — The three-point revolution

**Question.** How did the three-point shot change from the first seasons with recorded attempts to 2022–23?

The `game` table has one row per game, with home and away team box scores. In 1981–82, the database has 943 regular-season games, but only 113 (12.0%) have three-point-attempt values. Two of those 113 games are missing a field-goal-attempt value, so the shot-share calculation uses the 111 games with all four home/away 3PA/FGA values.

**Finding.** In the recorded 3PA rows, average attempts rose from 2.3 per team-game in 1981–82 to 34.2 in 2022–23. On the matched complete rows, threes rose from 2.6% to 38.7% of all field-goal attempts. Treat the early-season comparison cautiously because its coverage is incomplete. The 2012–13 season is absent from this database; the chart marks it as a gap rather than drawing a line through it.


## SQL and row coverage

The full reusable query is in `queries/insight_2.sql`. It uses a season calendar from 1981–82 through 2022–23, so the missing 2012–13 season remains an explicit row with null metrics. Three-point volume uses games with both 3PA values. Shot share separately filters to games with both home and away 3PA and FGA values, ensuring its numerator and denominator use the same rows. The coverage line reports games with both 3PA values divided by all games recorded for that season.


```python
from pathlib import Path
from IPython.display import display
from nba.data_loader import connect_database
from nba.insight_2 import three_point_summary, three_point_trend
from nba.insight_2_viz import plot_three_point_trend

connection = connect_database(Path('data/nba.sqlite'))
trend = three_point_trend(connection)
summary = three_point_summary(trend)
connection.close()

display(trend.head())
display(trend.tail())
display(summary)
```

```python
figure = plot_three_point_trend(trend)
display(figure)
```

## Interpretation and limitations

Attempts per team-game show three-point volume, while shot share measures those attempts against all field-goal attempts. Both measures rise strongly in the available records. The first season is a partial sample: the shot-share estimate uses 111 of 943 games, so coverage may affect how representative that early value is. The grey band marks the missing 2012–13 season; its null row breaks the lines. The database ends at 2022–23, so this is not the current season.
